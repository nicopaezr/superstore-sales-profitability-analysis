### **II. Data Preparation**

The dataset was prepared to ensure that variables are stored using appropriate data types required for the analysis.

```python
data["Order Date"] = pd.to_datetime(data["Order Date"], format = "%m/%d/%Y")
data["Ship Date"] = pd.to_datetime(data["Ship Date"], format = "%m/%d/%Y")
```

The `Order Date` and `Ship Date` variables were converted from the object data type to datetime to support analysis.

```python
data["Postal Code"] = data["Postal Code"].astype(str)
```

The `Postal Code` variable was converted from a numeric data type to object so it is treated as an identifier rather than a numerical measurement.

```python
# Extract the year into a new column
data["Year"] = data["Order Date"].dt.year
```

The `Year` variable was created to support time-based analysis of business performance. 

### **III. Overall Performance Analysis**

The overall sales and profitability of the superstore were examined to establish a baseline for business performance.


```python
total_sales = data["Sales"].sum()
print(total_sales)
```

The superstore had `total sales of $2,297,200.86` across all orders in the dataset. 

```python
total_profit = data["Profit"].sum()
print(total_profit)
```

The superstore generated a `total profit of $286,397.02` across all orders in the dataset.

```python
profit_margin = (total_profit / total_sales)*100
print(profit_margin)
```

The superstore had an overall profit margin of approximately `12.47%`. Therefore, about $12.47 of profit was generated for every $100 in sales. 

```python
number_orders = data["Order ID"].nunique()
print("Number of unique orders: ", number_orders)
number_customers = data["Customer ID"].nunique()
print("Number of unique customers: ", number_customers)
number_products = data["Product ID"].nunique()
print("Number of unique products sold: ", number_products)
```

The dataset represents `5,009` unique orders from `793` unique customers, covering `1,862` unique products.
